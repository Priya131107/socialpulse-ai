# SocialPulse AI — 01: Exploratory Data Analysis

This notebook analyzes the **SocialPulse AI** dataset (20,000 social posts) to uncover:
1. **Engagement Tier Imbalance** (Low vs Moderate vs Viral)
2. **Temporal Dynamics** (Hour-of-day vs normalized engagement)
3. **Platform & Brand Sentiment Breakdown**
4. **Brand Crisis Outliers & Anomaly Episodes**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Load processed training data
df = pd.read_csv('../data/processed/train.csv')
print(f'Train shape: {df.shape}')
df.head(3)
```

## 1. Class Distribution Analysis
Examining class proportions across Low (50%), Moderate (35%), and Viral (15%).

```python
class_counts = df['engagement_label'].value_counts(normalize=True) * 100
print(class_counts.to_string())
```

## 2. Temporal Engagement Patterns
Investigating how time of posting correlates with viral probability.

```python
df['timestamp'] = pd.to_datetime(df['timestamp'])
df['hour'] = df['timestamp'].dt.hour
hourly = df.groupby(['hour', 'engagement_label']).size().unstack().fillna(0)
hourly.plot(kind='bar', stacked=True, figsize=(12, 5), colormap='viridis')
plt.title('Hourly Post Counts by Engagement Tier')
plt.xlabel('Hour of Day')
plt.ylabel('Posts Count')
plt.show()
```
